## 2. Contract Type and Customer Churn

Contract type is an important factor associated with customer churn.

During exploratory data analysis, we observed differences in churn rates across contract categories.

The objective of this analysis is to understand how contract duration relates to customer retention and identify possible business strategies.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv("../data/processed/cleaned_telco_data.csv")

# Calculate churn percentage by contract type
contract_churn = (
    df.groupby("Contract")["Churn"]
    .value_counts(normalize=True)
    .unstack() * 100
)

# Extract churn percentage
contract_churn_rate = contract_churn["Yes"]

print("Churn Rate by Contract Type:")
print(contract_churn_rate.round(2))
```

```python
# Visualize churn rate by contract type

plt.figure(figsize=(8, 5))

contract_churn_rate.plot(kind="bar")

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 2.1 Business Insights

The analysis shows significant differences in churn rates across contract types.

- Month-to-month contracts have the highest churn rate of 42.71%.
- One-year contracts have a churn rate of 11.27%.
- Two-year contracts have the lowest churn rate of 2.83%.

**Business Recommendation:**

The telecom company can encourage month-to-month customers to consider longer-term contracts by providing attractive discounts, loyalty benefits and additional services.

This may help improve customer retention and reduce churn.

- Automatic credit card users have the lowest churn rate of 15.24%.

**Business Recommendations:**

1. Encourage customers using electronic checks to consider automatic payment options.
2. Simplify the payment process to improve customer convenience.
3. Monitor electronic check users for other signs of dissatisfaction.
4. Provide convenient and reliable billing options.

These strategies may help improve payment convenience and support customer retention.

## 7. Customer Tenure and Customer Churn

Customer tenure represents the number of months a customer has been associated with the telecom company.

Understanding tenure is important because customer behaviour may differ between new customers and long-term customers.

The objective of this analysis is to compare customer tenure between customers who churned and those who stayed.

```python
# Calculate average tenure based on churn status
tenure_analysis = df.groupby("Churn")["tenure"].mean()

print("Average Customer Tenure:")
print(tenure_analysis.round(2))
```

```python
# Visualize average tenure by churn status

plt.figure(figsize=(7, 5))

tenure_analysis.plot(kind="bar")

plt.title("Average Customer Tenure by Churn Status")
plt.xlabel("Customer Churn")
plt.ylabel("Average Tenure (Months)")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 7.1 Business Insights

The analysis reveals a difference in average tenure between customers who stayed and customers who churned.

- Customers who stayed had an average tenure of 37.57 months.
- Customers who churned had an average tenure of 17.98 months.

**Business Recommendations:**

1. Introduce onboarding programs for newly acquired customers.
2. Provide special offers during the initial months of service.
3. Regularly collect feedback from new customers.
4. Identify early signs of dissatisfaction and address them promptly.
5. Introduce loyalty benefits to encourage long-term customer relationships.

These strategies may help improve early customer engagement and support long-term retention.
